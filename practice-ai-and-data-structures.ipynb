# 1. Text Data and Unique Values

The first exercise works with descriptions of different AI models:

```python
model_descriptions = "ResNet is a convolutional neural network..."
```

Although this is initially stored as one long string, I need to transform it into a structure that Python can analyze.

### Step 1: Splitting text

The `.split()` method separates a string into individual words.

```python
model_list = model_descriptions.split()
```

For example:

```python
text = "AI models process data"
words = text.split()
```

produces:

```python
['AI', 'models', 'process', 'data']
```

This converts the original string into a **list**.

### Why it matters
Text data in NLP frequently starts as strings. Splitting text into smaller units is one of the basic steps in text preprocessing.

The resulting list can then be searched, counted, filtered, or transformed.

---

# 2. Using Sets to Find Unique Words

After creating the list of words, the next step is to remove duplicates.

```python
model_set = set(model_list)
```

A set stores **unique elements**, so repeated words such as:

```text
is
a
convolutional
neural
network
```

only appear once.

We can then determine how many unique words exist:

```python
num_unique = len(model_set)
```

The result for this exercise is:

```text
36
```

### Important distinction

A list:

```python
['AI', 'AI', 'ML', 'data']
```

can contain duplicates.

A set:

```python
{'AI', 'ML', 'data'}
```

contains only unique values.

### AI/NLP connection

This concept is useful when analyzing vocabulary.

For example, when processing a collection of documents, we may want to determine:

* How many unique words occur?
* Which words appear in the dataset?
* Which words are duplicated?
* How large is the vocabulary?

This is a basic concept that eventually connects to more advanced NLP techniques such as **tokenization, vocabulary construction, feature extraction, and text vectorization**.

# 3. Dictionaries for Model Performance

The second exercise introduces a different type of data:

```python
model_accuracy_dict = {
    'ResNet': 0.91,
    'MobileNet': 0.89,
    'VGG': 0.88,
    'Inception': 0.92,
    'AlexNet': 0.85,
    'EfficientNet': 0.93,
    'SqueezeNet': 0.87
}
```

This is a **dictionary**.

A dictionary stores information as:

```text
key → value
```

In this example:

```text
model name → accuracy
```

For example:

```python
model_accuracy_dict['ResNet']
```

returns:

```text
0.91
```

This is a natural representation for AI model evaluation because each model can be associated with a specific metric.

---

# 4. Counting Dictionary Entries

The `len()` function can be used to determine how many models are stored:

```python
num_models = len(model_accuracy_dict)
```

The result is:

```text
7
```

This demonstrates an important general Python principle:

> `len()` returns the number of elements in a container.

It can be used with many Python data structures, including:

```python
len(list)
len(set)
len(dictionary)
len(string)
```

For dictionaries, `len()` returns the number of **keys**.

---

# 5. Checking Whether a Key Exists

Python allows us to check whether a particular key exists using the `in` operator:

```python
contains_squeezenet = 'SqueezeNet' in model_accuracy_dict
```

The result is:

```text
True
```

This is preferable to manually inspecting the dictionary.

The general pattern is:

```python
key in dictionary
```

For example:

```python
'ResNet' in model_accuracy_dict
```

returns:

```text
True
```

while:

```python
'BERT' in model_accuracy_dict
```

would return:

```text
False
```

### Why this matters

In real applications, data may change dynamically. Instead of assuming that a particular key exists, a program can check first.

This is useful when working with:

* Model metadata
* Configuration files
* API responses
* JSON data
* Database records
* Dataset information

---

# 6. Sorting Dictionary Keys

The exercise asks us to sort the model names alphabetically.

The `sorted()` function can be used for this:

```python
sorted_keys = sorted(model_accuracy_dict.keys())
```

The result is a list containing the dictionary keys in sorted order.

Conceptually:

```text
AlexNet
EfficientNet
Inception
MobileNet
ResNet
SqueezeNet
VGG
```

The first element can then be accessed using index `0`:

```python
first_key = sorted_keys[0]
```

The result is:

```text
AlexNet
```

### Important concept

A dictionary and a list serve different purposes.

The dictionary is useful for mapping:

```text
model → accuracy
```

The sorted list is useful when we need an ordered collection of those model names.

This demonstrates how different Python data structures can work together.

---

# 7. Finding the Model with the Highest Accuracy

The final task is to determine which model has the highest accuracy.

Instead of manually inspecting every value, Python can perform the operation:

```python
highest_accuracy_model = max(
    model_accuracy_dict,
    key=model_accuracy_dict.get
)
```

The result is:

```text
EfficientNet
```

because:

```text
EfficientNet → 0.93
```

is the highest accuracy in the dictionary.

### Understanding `max()`

Normally, `max()` can find the largest value:

```python
numbers = [10, 25, 15]
max(numbers)
```

returns:

```text
25
```

But here we want the **key associated with the largest value**, rather than simply the largest key.

That's why we provide:

```python
key=model_accuracy_dict.get
```

This tells Python to compare the dictionary keys based on their corresponding values.

Conceptually, Python evaluates:

```text
ResNet       → 0.91
MobileNet    → 0.89
VGG          → 0.88
Inception    → 0.92
AlexNet      → 0.85
EfficientNet → 0.93  ← highest
SqueezeNet   → 0.87
```

and returns:

```text
EfficientNet
```

For example:

> "I can use Python's `max()` function with the dictionary's `get()` method as the key function, which allows Python to compare the accuracy values while returning the corresponding model name."

> "A list is ordered and can contain duplicates, while a set contains unique elements and is primarily useful when uniqueness and efficient membership testing are important."

## Quiz: Model Accuracy Dictionary

In the code cell below, you'll find a dictionary containing the names of different AI models as keys and their corresponding accuracy values stored as values. Use this dictionary to answer the following questions. Submit these answers in the quiz below this Jupyter Notebook.

````python
model_accuracy_dict = {
    'ResNet': 0.91, 'MobileNet': 0.89, 'VGG': 0.88, 'Inception': 0.92, 
    'AlexNet': 0.85, 'EfficientNet': 0.93, 'SqueezeNet': 0.87
}

print(model_accuracy_dict, '\n')

# find number of unique keys in the dictionary
num_models = len(model_accuracy_dict)
print(num_models)

# find whether 'SqueezeNet' is a key in the dictionary
contains_squeezenet = 'SqueezeNet' in model_accuracy_dict
print(contains_squeezenet)

# create and sort a list of the dictionary's keys
sorted_keys = sorted(model_accuracy_dict.keys())
print(sorted_keys)

# get the first element in the sorted list of keys
first_key = sorted_keys[0]
print(first_key)

# find the element with the highest value in the dictionary
highest_accuracy_model = max(model_accuracy_dict, key=model_accuracy_dict.get)
print(highest_accuracy_model)
````

# 1. Python Variables and Assignment Operators

## What Is a Variable?

A variable is a name that refers to a value in a Python program.

For example:

```python
learning_rate = 0.01
```

Here:

* `learning_rate` is the variable name.
* `0.01` is the value.
* `=` assigns the value to the variable.

Variables allow programs to store information and update it as needed.

### Why Variables Matter in AI/ML

Machine learning programs use variables to represent different types of information, including:

| Variable        | Meaning in AI/ML                                                            |
| --------------- | --------------------------------------------------------------------------- |
| `learning_rate` | A hyperparameter controlling the size of model updates during optimization  |
| `training_loss` | A value representing the model's error according to a loss function         |
| `epochs`        | The number of complete passes through the training dataset                  |
| `batch_size`    | The number of training samples processed in one batch                       |
| `accuracy`      | A measure of the proportion of correct predictions in a classification task |

These variables help developers and data scientists manage model settings and track results.

---

## Understanding Assignment Operators

Assignment operators are used to assign or update values in Python.

The basic assignment operator is:

```python
=
```

Example:

```python
learning_rate = 0.01
```

Python also provides compound assignment operators.

### Common Assignment Operators

| Operator | Meaning             | Example  | Equivalent Expression |
| -------- | ------------------- | -------- | --------------------- |
| `=`      | Assign a value      | `x = 10` | `x = 10`              |
| `+=`     | Add and assign      | `x += 2` | `x = x + 2`           |
| `-=`     | Subtract and assign | `x -= 2` | `x = x - 2`           |
| `*=`     | Multiply and assign | `x *= 2` | `x = x * 2`           |
| `/=`     | Divide and assign   | `x /= 2` | `x = x / 2`           |

### Example: Updating a Variable

```python
accuracy = 85.5

accuracy += 2.0

print(accuracy)
```

Output:

```text
87.5
```

The `+=` operator adds `2.0` to the existing value.

This same programming concept can be used to update values in AI/ML-related programs.

# 2. Understanding Hyperparameters in Machine Learning

## What Are Hyperparameters?

Hyperparameters are settings chosen before or around the training of a machine learning model. They influence how the model learns or how the training process is configured.

They are different from model parameters.

### Model Parameters vs. Hyperparameters

| Model Parameters                        | Hyperparameters                                |
| --------------------------------------- | ---------------------------------------------- |
| Learned from training data              | Set by the developer or training configuration |
| Updated during model training           | Used to control the training process           |
| Examples include neural network weights | Examples include learning rate and batch size  |

For example, a neural network learns weights during training, while a developer may choose the learning rate before training begins.

### Common Hyperparameters

#### 1. Learning Rate

Controls the step size used by an optimization algorithm when updating model parameters to reduce the loss function.

A learning rate that is too large may cause unstable training. A learning rate that is too small may make training slow.

#### 2. Batch Size

Determines how many training samples are processed before a model update.

Example:

```python
batch_size = 32
```

This means a batch contains 32 samples, assuming the training configuration uses that batch size.

#### 3. Number of Epochs

Specifies how many times the training algorithm passes through the entire training dataset.

Example:

```python
epochs = 10
```

This represents 10 complete passes through the training dataset.

#### 4. Regularization Parameters

Regularization settings help control model complexity and can reduce overfitting.

Examples include L1 and L2 regularization strengths.

#### 5. Momentum

Momentum is an optimization setting used by some algorithms to help influence the direction and speed of parameter updates.

# 3. Understanding Training Loss

## What Is Training Loss?

Training loss is a numerical measure of how far a model's predictions are from the desired targets according to a chosen loss function.

A loss function measures prediction error during training.

For many optimization problems, the goal is to reduce the loss over time.

For example:

```python
training_loss = 0.8
```

This stores a simplified loss value of `0.8`.

### Important Note

The meaning of a loss value depends on the loss function being used.

For example:

* Mean Squared Error measures squared prediction errors.
* Cross-entropy loss measures a type of difference between predicted probabilities and target labels.

A lower loss is often desirable during training, but loss values should be interpreted in the context of the model, task, and loss function.

# 5. Learning Rate and Scientific Notation

## What Is a Learning Rate?

The learning rate is an important hyperparameter used by optimization algorithms such as gradient descent.

It controls how large the update to model parameters can be during optimization.

A simplified update rule can be written as:

$$
\theta_{\text{new}} =
\theta_{\text{old}} -
\eta \nabla L(\theta)
$$

Where:

* \(\theta\) represents model parameters.
* \(\eta\) represents the learning rate.
* \(L(\theta)\) represents the loss function.
* \(\nabla L(\theta)\) represents the gradient of the loss.

The learning rate determines the size of the step taken in the direction that aims to reduce the loss.

### Example

```python
learning_rate = 0.001
```

A smaller learning rate generally means smaller parameter updates.

However, the best learning rate depends on the model, dataset, optimizer, and training problem.

---

## Scientific Notation in Python

Learning rates are often small numbers, so scientific notation is useful.

For example:

```python
learning_rate = 1e-3
```

This is equivalent to:

```python
learning_rate = 0.001
```

### Understanding the Notation

In Python:

```python
1e-3
```

means:

$$
1 \times 10^{-3} = 0.001
$$

Other examples:

| Scientific Notation | Decimal |
| ------------------- | ------: |
| `1e-1`              |     0.1 |
| `1e-2`              |    0.01 |
| `1e-3`              |   0.001 |
| `1e-4`              |  0.0001 |
| `5e-4`              |  0.0005 |

````python
# The initial learning rate for the model
learning_rate = 0.01
# The initial training loss of the model
training_loss = 0.8

# Use the multiplication assignment operator to decrease 
# the learning_rate variable by 10% to account for better stability
learning_rate *= 0.9

### Notebook grading
if learning_rate == 0.01 * 0.9:
    print("Nice job! You can use the multiplication assignment operator like this: learning_rate *= 0.9")
elif learning_rate == 0.01 * 1.1:
    print("Sorry! learning_rate is too high. It looks like you increased the learning rate by 10% instead of decreasing it. Try again")        
else:
    print("Sorry! learning_rate is incorrect. Try again and make sure to use the *= assignment operator")
````

````python
# Use the addition assignment operator to add 
# a small value (0.02) to the training_loss variable due to noisy data
training_loss += 0.02

### Notebook grading
if training_loss == 0.8 + 0.02:
    print("Nice job! You can use the addition assignment operator like this: training_loss += 0.02")
elif training_loss == 0.8:
    print("Sorry! training_loss is too low. Did you add the noise? Try again")        
else:
    print("Sorry! training_loss is incorrect. Try again and make sure to use the += assignment operator")
````

````python
# Use the multiplication assignment operator to increase 
# training_loss by 5% to account for overfitting
training_loss *= 1.05

### Notebook grading
if training_loss == (0.8 + 0.02) * 1.05:
    print("Nice job! You can use the multiplication assignment operator like this: training_loss *= 1.05")
elif training_loss == 0.8 + 0.02:
    print("Sorry! training_loss is too low. Did you account for overfitting? Try again")        
else:
    print("Sorry! training_loss is incorrect. Try again and make sure to use the *= assignment operator")
````

````python
# Use the multiplication assignment operator to decrease 
# training_loss by 5% to account for regularization
training_loss *= 0.95

### Notebook grading
if training_loss == (0.8 + 0.02) * 1.05 * 0.95:
    print("Nice job! You can use the multiplication assignment operator like this: training_loss *= 0.95")
elif training_loss == (0.8 + 0.02) * 1.05:
    print("Sorry! training_loss is too high. Did you account for regularization? Try again")        
else:
    print("Sorry! training_loss is incorrect. Try again and make sure to use the *= assignment operator")
````

````python
# Use the subtraction assignment operator to subtract 
# a small value (0.05) from training_loss to account 
# for improved generalization due to more training data.
training_loss -= 0.05

### Notebook grading
if training_loss == (0.8 + 0.02) * 1.05 * 0.95 - 0.05:
    print("Nice job! You can use the subtraction assignment operator like this: training_loss -= 0.05")
elif training_loss == (0.8 + 0.02) * 1.05 * 0.95:
    print("Sorry! training_loss is too high. Did you account for the improved generalization? Try again")        
else:
    print("Sorry! training_loss is incorrect. Try again and make sure to use the -= assignment operator")
````

### Understanding Hyperparameters: Learning Rate
Learning Rate is a crucial hyperparameter in machine learning models that determines the step size at each iteration while moving towards a minimum of the loss function. It is common to use [scientific notation](https://en.wikipedia.org/wiki/Scientific_notation) for learning rates because they are often very small values. For example, a learning rate of `0.001` can be represented as `1𝑒−3`.

````python
# Initial learning rate
learning_rate = 1e-3  # Equivalent to 0.001

# Update the learning rate by decreasing it by 10%
learning_rate *= 0.9

print(f"Updated learning rate: {learning_rate}")
````

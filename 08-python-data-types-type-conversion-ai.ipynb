# 1. Understanding Python Data Types

A **data type** describes the kind of value stored in a Python variable.

Some common Python data types include:

| Data Type | Example          | Description          |
| --------- | ---------------- | -------------------- |
| `int`     | `10`             | Whole numbers        |
| `float`   | `0.15`           | Decimal numbers      |
| `str`     | `"0.15"`         | Text                 |
| `bool`    | `True`           | Boolean values       |
| `list`    | `[1, 2, 3]`      | Ordered collection   |
| `dict`    | `{"loss": 0.15}` | Key-value collection |

For example:

```python
integer_value = 10
decimal_value = 0.15
text_value = "0.15"
boolean_value = True
```

We can inspect the type of a variable using the `type()` function:

```python
print(type(integer_value))
print(type(decimal_value))
print(type(text_value))
print(type(boolean_value))
```

This helps us understand how Python will treat each value.

---

# 2. Strings vs Numerical Values

A particularly important distinction in this exercise is the difference between a numerical value and a string containing a number.

Consider:

```python
loss = 0.15
```

Here, `loss` is a floating-point number:

```python
type(loss)
```

returns:

```text
float
```

However:

```python
loss = "0.15"
```

stores the value as text.

Its type is:

```text
str
```

Even though `"0.15"` looks like a number, Python treats it as a string.

---

# 3. Type Conversion

**Type conversion** is the process of changing a value from one data type to another.

Python provides several built-in functions for converting values.

### Converting to an Integer

```python
number = int("10")
```

The result is:

```python
10
```

with type:

```text
int
```

### Converting to a Float

```python
number = float("0.15")
```

The result is:

```text
0.15
```

with type:

```text
float
```

### Converting to a String

```python
number = 100
text = str(number)
```

The result is the string:

```text
"100"
```

Type conversion is particularly important when data comes from external sources.

---

# 4. Why Type Conversion Matters in AI and Machine Learning

In real-world AI applications, data does not always arrive in the format we expect.

For example, a CSV file might contain:

```text
loss
0.15
0.12
0.13
```

Depending on how the data is loaded or processed, these values may initially be represented as strings.

If we attempt to perform mathematical operations on strings, Python may produce an error or perform an operation that does not represent the intended calculation.

For example:

```python
loss1 = "0.15"
loss2 = "0.12"

# This does not perform numerical addition.
result = loss1 + loss2
```

The result would be:

```text
"0.150.12"
```

because Python is concatenating two strings.

To perform numerical addition, we need to convert the values:

```python
result = float(loss1) + float(loss2)
```

The result is:

```text
0.27
```

This demonstrates why understanding data types is essential when preparing data for machine learning workflows.

# 5. Understanding Loss in Machine Learning

**Loss** is a numerical measure of how different a model's predictions are from the expected target values.

During model training, a loss function evaluates the model's predictions and produces a numerical value representing the error.

In general:

> Lower loss values indicate that the model's predictions are closer to the expected targets.

For many machine learning algorithms, the objective of training is to minimize the loss.

For example, a model might produce the following loss values during training:

```text
Epoch 1 → 0.50
Epoch 2 → 0.40
Epoch 3 → 0.30
Epoch 4 → 0.20
Epoch 5 → 0.15
```

The decreasing values suggest that the model is improving according to the selected loss function.

However, the interpretation of a loss value depends on the specific loss function and problem being solved.

---

# 6. Aggregating Performance Metrics

A model may produce performance measurements at different points during training.

For example:

```text
Monday    → 0.15
Tuesday   → 0.12
Wednesday → 0.13
Thursday  → 0.10
Friday    → 0.11
Saturday  → 0.09
Sunday    → 0.08
```

We can use Python to aggregate these values.

For example, the total can be calculated as:

```python
total_loss = (
    0.15 + 0.12 + 0.13 +
    0.10 + 0.11 + 0.09 + 0.08
)
```

The resulting total is:

```text
0.78
```

Although a total loss can be useful for this exercise, in real machine learning evaluation we would often use other summaries such as:

* Mean loss
* Minimum loss
* Maximum loss
* Loss over time
* Validation loss
* Training loss

The choice depends on what we are trying to understand about the model.

---

# 7. Formatting Numerical Results

After calculating a numerical result, we may want to present it in a readable format.

Python's `.format()` method can be used for this purpose.

For example:

```python
total_loss = 0.7765

result = "This week's total loss: {:.2f}".format(total_loss)
```

The `:.2f` formatting instruction means that the number should be displayed with two decimal places.

The result would be:

```text
This week's total loss: 0.78
```

Useful when presenting numerical results in reports, dashboards, logs, or user-facing applications.

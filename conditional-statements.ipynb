# Python Conditional Statements
## 1. Introduction

Conditional statements are one of the fundamental building blocks of programming. They allow a program to **make decisions based on information available at runtime**.

In real programs, we rarely want every line of code to execute regardless of the situation. Instead, we often need the program to ask questions such as:

* Is the user's balance low?
* Is the model accuracy above the required threshold?
* Is the input valid?
* Does a particular record exist?
* Is the prediction confidence high enough?
* Should a particular action be performed?
* Does a transaction meet a certain condition?

Python uses `if`, `elif`, and `else` statements to implement this type of decision-making.

For AI and data science, conditional statements are particularly important because data and model outputs often need to be evaluated before the program decides what to do next.

For example:

```python
accuracy = 0.92

if accuracy >= 0.90:
    print("Model meets the required accuracy.")
```

Here, Python evaluates the condition:

```python
accuracy >= 0.90
```

The result is either `True` or `False`.

Since `0.92 >= 0.90` is `True`, Python executes the indented statement.

This demonstrates an important programming concept:

> **A condition controls whether a particular block of code executes.**

---

# 2. What Is an `if` Statement?

An `if` statement is a conditional statement that allows Python to execute a block of code **only when a specified condition evaluates to `True`**.

The general structure is:

```python
if condition:
    # code to execute if condition is True
```

For example:

```python
phone_balance = 3
bank_balance = 100

if phone_balance < 5:
    phone_balance += 10
    bank_balance -= 10

print(phone_balance)
print(bank_balance)
```

### What happens here?

Python evaluates:

```python
phone_balance < 5
```

Since:

```python
3 < 5
```

is `True`, Python executes:

```python
phone_balance += 10
bank_balance -= 10
```

The resulting values are:

```text
13
90
```

If `phone_balance` had been `10`, the condition would have been `False`, and Python would simply skip the indented block.

---

# 3. Understanding the Structure of an `if` Statement

An `if` statement contains several important components.

```python
if condition:
    statement
```

### `if`

The `if` keyword tells Python that we are about to make a decision.

### Condition

The condition is an expression that Python evaluates.

For example:

```python
phone_balance < 5
```

The result will be either:

```python
True
```

or:

```python
False
```

### Colon `:`

The colon tells Python that the conditional block begins after this line.

```python
if phone_balance < 5:
```

### Indented block

The indented statements belong to the `if` block.

```python
if phone_balance < 5:
    phone_balance += 10
    bank_balance -= 10
```

Both statements are executed only when the condition is `True`.

---

# 4. Boolean Expressions and Conditional Logic

Conditional statements depend heavily on **Boolean expressions**.

A Boolean expression produces one of two values:

```python
True
False
```

For example:

```python
accuracy = 0.91

print(accuracy >= 0.90)
```

Output:

```text
True
```

Another example:

```python
accuracy = 0.84

print(accuracy >= 0.90)
```

Output:

```text
False
```

This is the foundation of program decision-making.

A useful way to think about an `if` statement is:

```text
Evaluate condition
       ↓
Is it True?
   ↙       ↘
 Yes        No
 ↓           ↓
Run block   Skip block
```

---

# 5. Comparison Operators

Comparison operators are frequently used inside conditional statements.

| Operator | Meaning                  | Example              |
| -------- | ------------------------ | -------------------- |
| `==`     | Equal to                 | `accuracy == 0.90`   |
| `!=`     | Not equal to             | `model != "Unknown"` |
| `>`      | Greater than             | `accuracy > 0.90`    |
| `<`      | Less than                | `accuracy < 0.90`    |
| `>=`     | Greater than or equal to | `accuracy >= 0.90`   |
| `<=`     | Less than or equal to    | `accuracy <= 0.90`   |

For example:

```python
accuracy = 0.93

if accuracy >= 0.90:
    print("The model meets the target.")
```

Output:

```text
The model meets the target.
```

---

# 6. `==` vs `=`

One of the most important distinctions to understand is the difference between:

```python
=
```

and:

```python
==
```

They have completely different purposes.

## Assignment operator: `=`

The single equals sign assigns a value to a variable.

```python
accuracy = 0.92
```

This means:

> Store `0.92` inside the variable `accuracy`.

## Comparison operator: `==`

The double equals sign compares two values.

```python
accuracy == 0.92
```

This asks:

> Is the value of `accuracy` equal to `0.92`?

The result is Boolean:

```python
True
```

or:

```python
False
```

### Example

Correct:

```python
if accuracy == 0.92:
    print("Accuracy is exactly 92%.")
```

Incorrect:

```python
if accuracy = 0.92:
    print("Accuracy is exactly 92%.")
```

The second version produces a syntax error because Python does not allow assignment in that position.

### Key distinction

```text
=   → assign a value
==  → compare values
```

This distinction is extremely important in Python programming and technical interviews.

---

# 7. The `else` Clause

Sometimes we need the program to do something when the condition is `False`.

This is where `else` is useful.

```python
accuracy = 0.84

if accuracy >= 0.90:
    print("Model meets the target.")
else:
    print("Model needs improvement.")
```

Output:

```text
Model needs improvement.
```

The structure is:

```python
if condition:
    # runs when condition is True
else:
    # runs when condition is False
```

Unlike `if`, `else` does not require its own condition.

It represents the alternative case.

---

# 8. Using `elif`

Sometimes there are more than two possible situations.

For example, suppose we want to classify model performance:

```python
accuracy = 0.87

if accuracy >= 0.90:
    print("Excellent")
elif accuracy >= 0.80:
    print("Good")
else:
    print("Needs improvement")
```

Output:

```text
Good
```

`elif` means:

> Check another condition if the previous condition was `False`.

We can have multiple `elif` statements:

```python
if condition_1:
    ...
elif condition_2:
    ...
elif condition_3:
    ...
else:
    ...
```

Python checks the conditions **from top to bottom**.

---

# 9. How Python Chooses a Branch

Consider:

```python
score = 75

if score >= 90:
    print("Excellent")
elif score >= 80:
    print("Very Good")
elif score >= 70:
    print("Good")
else:
    print("Needs Improvement")
```

Python evaluates the conditions in order.

### First:

```python
score >= 90
```

`75 >= 90` → `False`

### Second:

```python
score >= 80
```

`75 >= 80` → `False`

### Third:

```python
score >= 70
```

`75 >= 70` → `True`

Therefore Python prints:

```text
Good
```

Once Python finds a `True` condition in an `if`/`elif` chain, it executes that block and skips the remaining branches.

This is important because **the order of conditions can affect the result**.

# 13. Indentation in Python

One of the most important features of Python is that **indentation defines code blocks**.

For example:

```python
accuracy = 0.92

if accuracy >= 0.90:
    print("Model accepted")
```

The `print()` statement is indented, so Python knows that it belongs to the `if` block.

Compare this with:

```python
accuracy = 0.92

if accuracy >= 0.90:
print("Model accepted")
```

This produces an `IndentationError`.

Python expects the code belonging to the `if` statement to be indented.

---

# 14. Why Indentation Matters

Consider:

```python
if accuracy >= 0.90:
    print("Model accepted")
    print("Ready for further testing")
```

Both statements belong to the `if` block.

But:

```python
if accuracy >= 0.90:
    print("Model accepted")

print("Evaluation complete")
```

Only the first statement belongs to the `if`.

The final `print()` executes regardless of whether the accuracy condition is true or false.

Therefore, indentation is not merely formatting in Python.

**Indentation affects program behavior.**

---

# 15. Python's Four-Space Convention

The standard Python convention is to use **four spaces** for indentation.

For example:

```python
if accuracy >= 0.90:
    print("Accepted")
```

If another block is nested inside it, another level of indentation is added:

```python
if accuracy >= 0.90:
    if confidence >= 0.95:
        print("High-confidence model prediction")
```

Here:

* First level → 4 spaces
* Second level → 8 spaces

Consistent indentation makes code easier to understand and prevents errors.

---

# 16. Spaces vs Tabs

Python code should generally follow the standard four-space indentation convention.

Avoid mixing tabs and spaces within the same program.

A good development environment such as VS Code will normally handle indentation automatically, but it is still important to understand what Python is doing.

The important principle is:

> **Indentation is part of Python syntax, not just visual formatting.**

# 17. Nested Conditional Statements

An `if` statement can also appear inside another `if` statement.

For example:

```python
accuracy = 0.94
confidence = 0.97

if accuracy >= 0.90:
    print("Model meets accuracy target.")

    if confidence >= 0.95:
        print("Prediction confidence is also high.")
```

Here the second condition is only checked after the first condition is satisfied.

This can be useful, but deeply nested conditions can make code difficult to read. As programs become more complex, it is often better to simplify the logic.

---

# 18. Multiple Conditions

Python allows conditions to be combined using logical operators.

### `and`

Both conditions must be `True`.

```python
accuracy = 0.94
confidence = 0.96

if accuracy >= 0.90 and confidence >= 0.95:
    print("Model meets both requirements.")
```

### `or`

At least one condition must be `True`.

```python
if accuracy >= 0.90 or confidence >= 0.95:
    print("Model meets at least one requirement.")
```

### `not`

Reverses a Boolean value.

```python
is_deployed = False

if not is_deployed:
    print("Model is not deployed.")
```

These operators become particularly useful when building real-world decision logic.

# 19. Common Mistakes I Need to Avoid

## Mistake 1: Using `=` instead of `==`

Incorrect:

```python
if model = "ResNet":
```

Correct:

```python
if model == "ResNet":
```

---

## Mistake 2: Forgetting the colon

Incorrect:

```python
if accuracy >= 0.90
    print("Accepted")
```

Correct:

```python
if accuracy >= 0.90:
    print("Accepted")
```

---

## Mistake 3: Incorrect indentation

Incorrect:

```python
if accuracy >= 0.90:
print("Accepted")
```

Correct:

```python
if accuracy >= 0.90:
    print("Accepted")
```

---

## Mistake 4: Incorrect ordering of conditions

Consider:

```python
score = 95

if score >= 70:
    print("Good")
elif score >= 90:
    print("Excellent")
```

The output is:

```text
Good
```

The `>= 90` condition is never reached because `95 >= 70` is already `True`.

A better ordering is:

```python
if score >= 90:
    print("Excellent")
elif score >= 70:
    print("Good")
else:
    print("Needs improvement")
```

This is an important lesson:

> **In an `if`/`elif` chain, condition order matters.**

# 23. Interview Preparation

### What is an `if` statement?

An `if` statement is a conditional statement that executes a block of code when a specified Boolean condition evaluates to `True`.

### What is the difference between `=` and `==`?

`=` is the assignment operator, while `==` is the equality comparison operator.

### What is the purpose of `elif`?

`elif` allows Python to check additional conditions when preceding conditions are false.

### Is `else` required?

No. An `if` statement can exist without `elif` or `else`.

### Why is indentation important in Python?

Python uses indentation to define code blocks, so incorrect indentation can change program behavior or produce an error.

### Can an `if` statement have multiple `elif` clauses?

Yes. Multiple `elif` clauses can be used to evaluate several possible conditions.

### Does Python execute every branch of an `if`/`elif`/`else` statement?

No. Once Python finds the first condition that evaluates to `True`, it executes that branch and skips the remaining branches.

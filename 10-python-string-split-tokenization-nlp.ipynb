# 2. Tokenization
**Tokenization** is the process of breaking text into smaller units called tokens.
A token may be:

* A word.
* A punctuation mark.
* A subword.
* A sentence.

For example:

```text
I love machine learning.
```

can be split into word tokens:

```python
["I", "love", "machine", "learning."]
```

Using Python:

```python
sentence = "I love machine learning."

tokens = sentence.split()

print(tokens)
```

Output:
```text
['I', 'love', 'machine', 'learning.']
```
The `.split()` method separates the string wherever whitespace occurs.

---

# 3. Counting Tokens
The `len()` function can be used to count the number of tokens in a list.
```python
sentence = "I love machine learning."

tokens = sentence.split()

print(len(tokens))
```

Output:

```text
4
```

This provides a simple way to estimate the number of word tokens in a sentence.

However, `.split()` is a basic approach. More advanced NLP tokenizers may handle punctuation, contractions, and subwords differently.

---

# 4. Finding Words in Text
Python can be used to check whether a word appears in a sentence.
For example:

```python
review = "The product is excellent and easy to use."

print("excellent" in review)
```

Output:

```text
True
```

The `in` operator checks whether a sequence of characters exists within a string.
To avoid missing words because of capitalization, the text can first be converted to lowercase:

```python
review = "The Product Is Excellent."

print("excellent" in review.lower())
```

Output:

```text
True
```

This is a simple form of text normalization.

---

# 5. Counting Word Occurrences
The `.count()` method can be used to count how many times a particular sequence appears in a string.
```python
text = "AI is useful. AI is powerful. AI is changing industries."

count = text.count("AI")

print(count)
```

Output:

```text
3
```

For more reliable word-frequency analysis, text is usually tokenized and normalized before counting.

For example:

```python
text = "AI is useful. AI is powerful. AI is changing industries."

tokens = text.lower().replace(".", "").split()

print(tokens.count("ai"))
```

Output:

```text
3
```

# 6. Sentiment Analysis
**Sentiment analysis** is an NLP task that attempts to identify the emotional tone or opinion expressed in text.

It is commonly used for:
* Customer reviews.
* Product feedback.
* Social media monitoring.
* Survey responses.
* Brand analysis.
* Customer support messages.

A sentiment analysis system may classify text as:
* Positive.
* Negative.
* Neutral.

For example:

```text
"The product is excellent and reliable."
```

may be classified as positive.

```text
"The service was slow and disappointing."
```

may be classified as negative.

---

# 7. Basic Rule-Based Sentiment Analysis
A simple sentiment-analysis approach uses predefined positive and negative words.

For example:

```python
positive_words = ["good", "great", "excellent", "love", "happy"]
negative_words = ["bad", "poor", "terrible", "hate", "disappointing"]
```

The text can then be converted to lowercase and split into tokens.

The number of positive and negative words can be counted to produce a basic sentiment result.

This approach is useful for learning, but it has limitations. It may not understand:

* Sarcasm.
* Negation.
* Context.
* Slang.
* Mixed opinions.
* Complex sentence structure.

For example:

```text
"The product is not good."
```

contains the word `good`, but the overall sentiment is negative.

Real sentiment-analysis systems often use machine learning, deep learning, or language models to understand context more effectively.

### Task 5: Basic Sentiment Analysis

Create two lists:

```python
positive_words = ["good", "great", "excellent", "love", "happy"]
negative_words = ["bad", "poor", "terrible", "hate", "disappointing"]
```

Convert the review to lowercase and identify whether it contains positive or negative words.

The review contains the word `"excellent"`, so a simple keyword-based system would classify it as positive.

---

## Example Solution

```python
review = "The product is excellent and easy to use."

# Tokenize the review
tokens = review.split()

# Count tokens
token_count = len(tokens)

# Normalize the text
normalized_review = review.lower()

# Check for a word
contains_excellent = "excellent" in normalized_review

# Count occurrences
excellent_count = tokens.count("excellent")

# Basic sentiment word lists
positive_words = ["good", "great", "excellent", "love", "happy"]
negative_words = ["bad", "poor", "terrible", "hate", "disappointing"]

# Count positive and negative words
positive_count = sum(
    word in normalized_review for word in positive_words
)

negative_count = sum(
    word in normalized_review for word in negative_words
)

# Determine basic sentiment
if positive_count > negative_count:
    sentiment = "Positive"
elif negative_count > positive_count:
    sentiment = "Negative"
else:
    sentiment = "Neutral"

print("Tokens:", tokens)
print("Token count:", token_count)
print("Contains 'excellent':", contains_excellent)
print("Excellent count:", excellent_count)
print("Positive word count:", positive_count)
print("Negative word count:", negative_count)
print("Basic sentiment:", sentiment)
```

## Quiz: Analyzing AI Model Output Text
### Scenario:
You are analyzing the output of an AI model from a recent project. You need to answer specific questions about the text using string methods.

### Task:
Answer the following questions about the string variable model_output:

1. What is the length of the string variable model_output?
2. What is the index of the first occurrence of the word 'accuracy' in model_output?
3. What is the index of the last occurrence of the word 'model' in model_output?
4. What is the count of occurrences of the word 'model' in model_output?
5. Tokenize the model_output into individual words.
6. Perform a basic sentiment analysis by counting the number of positive and negative words in the model_output.

````python
model_output = "The AI model achieved an accuracy of 92.5% in the initial tests.\nThe model's performance was consistent across different datasets.\nFurther tuning of the model hyperparameters improved the accuracy to 94%.\nThis model is now ready for deployment in the production environment.\nModel performance will be monitored continuously to ensure it meets the expected standards."

print(model_output)

# Answer the questions
length_of_model_output = len(model_output)
index_first_accuracy = model_output.find("accuracy")
index_last_model = model_output.find("model")
count_model = model_output.count("model")

# Tokenize the model_output into individual words
tokens = model_output.split()

# Perform a basic sentiment analysis
positive_words = ['achieved', 'consistent', 'improved', 'ready', 'meets']
negative_words = ['monitored']

count_positive = sum(model_output.lower().count(word) for word in positive_words)
count_negative = sum(model_output.lower().count(word) for word in negative_words)
print("-"*50)
# Output your answers in descriptive messages
print("The length of the model output string is {} characters.".format(length_of_model_output))
print("The first occurrence of the word 'accuracy' is at index {}.".format(index_first_accuracy))
print("The last occurrence of the word 'model' is at index {}.".format(index_last_model))
print("The word 'model' occurs {} times in the model output.".format(count_model))
print("The tokenized model output is: {}".format(tokens))
print("The number of positive words is: {}".format(count_positive))
print("The number of negative words is: {}".format(count_negative))
````

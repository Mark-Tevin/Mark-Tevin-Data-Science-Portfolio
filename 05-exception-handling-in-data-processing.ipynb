# 1. Understanding Exceptions in Python

## What Is an Exception?

An exception is an event that occurs during program execution when Python encounters an error or an unexpected condition.

For example:

```python
result = 100 / 0
```

This causes:

```text
ZeroDivisionError
```

Python raises this exception because division by zero is not allowed.

### Why Exceptions Matter

Without exception handling, an unexpected error can stop a program from completing its work.

In AI/ML projects, an error during data processing could interrupt a pipeline or prevent a model training process from continuing.

Exception handling provides a way to respond to known error conditions.

# 2. Common Python Exceptions

## ZeroDivisionError

Occurs when a number is divided by zero.

Example:

```python
result = 100 / 0
```

This raises a `ZeroDivisionError`.

### AI/ML Connection

A data processing pipeline may perform calculations involving numerical values. If an input value is zero and the operation requires division by that value, an exception may occur.

---

## ValueError

Occurs when a function receives a value of the correct general type, but the value is inappropriate for the operation.

Example:

```python
number = int("hello")
```

This raises a `ValueError` because `"hello"` cannot be converted into an integer.

### AI/ML Connection

This can happen when converting invalid text data into numerical values during data preprocessing.

---

## TypeError

Occurs when an operation is applied to an object of an inappropriate type.

Example:

```python
result = "100" / 2
```

This raises a `TypeError` because a string cannot be divided by an integer.

### AI/ML Connection

This may occur when a program expects numerical data but receives a string or another incompatible data type.

# 3. Understanding try and except

## The try Block

The `try` block contains code that may raise an exception.

Example:

```python
try:
    result = 100 / value
```

Python attempts to execute the code inside the block.

## The except Block

The `except` block defines what should happen if a specific exception occurs.

Example:

```python
except ZeroDivisionError:
    print("Cannot divide by zero.")
```

If the division causes a `ZeroDivisionError`, Python executes the exception handler.

### Basic Structure

```python
try:
    # Code that may cause an exception
    result = 100 / value

except ZeroDivisionError:
    # Code that handles the exception
    print("Cannot divide by zero.")
```

### How the Code Works

1. Python enters the `try` block.
2. It attempts to perform the division.
3. If the division succeeds, the result is printed.
4. If a `ZeroDivisionError` occurs, Python skips to the `except` block.
5. The friendly error message is printed.

This allows the program to respond to the known error condition.

# 4. Building the Data Processing Function

## Function Objective

The function `process_data(value)` will perform a division operation using the supplied input.

The function will demonstrate how exception handling can protect a calculation from division by zero.

### Function Structure

```python
def process_data(value):
    try:
        result = 100 / value
        print(f"Result: {result}")
    except ZeroDivisionError:
        print("Cannot process the data because division by zero is not allowed.")
```

### Explanation

* `def process_data(value):` defines a function.
* `value` is the input argument.
* `try` contains the division operation.
* `result` stores the calculation.
* `except ZeroDivisionError` catches division-by-zero errors.
* `print()` displays either the result or an error message.

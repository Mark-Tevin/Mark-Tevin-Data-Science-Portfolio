## Common String Methods

Some useful Python string methods include when learning:

| Method       | Purpose                        | Example                       |
| ------------ | ------------------------------ | ----------------------------- |
| `.lower()`   | Converts text to lowercase     | `"HELLO".lower()`             |
| `.upper()`   | Converts text to uppercase     | `"hello".upper()`             |
| `.replace()` | Replaces part of a string      | `"cat".replace("cat", "dog")` |
| `.split()`   | Splits text into a list        | `"hello world".split()`       |
| `.strip()`   | Removes surrounding whitespace | `" hello ".strip()`           |

These operations are commonly useful when cleaning and preparing text data.

# About BERT

**BERT** stands for **Bidirectional Encoder Representations from Transformers**.

BERT is a transformer-based language model introduced by Google and designed to understand the context of words by considering information from both directions of a sentence.

It has been widely used and adapted for NLP tasks such as:

* Text classification
* Named Entity Recognition
* Question answering
* Sentiment analysis
* Information extraction

The exercise does not train or use a BERT model directly. Instead, it uses BERT as an example of how model information can be represented and formatted using Python.

## Understanding Classification Metrics

Machine learning classification models make predictions about whether an observation belongs to a particular class.

Examples include:
- Fraudulent or legitimate transactions.
- Spam or non-spam emails.
- Positive or negative medical diagnoses.
- Defective or non-defective products.
- Malicious or normal network activity.

To evaluate these predictions, we use values from the confusion matrix.

### True Positive

A **True Positive (TP)** occurs when the model correctly predicts a positive instance.

Example:
A transaction is fraudulent, and the model correctly identifies it as fraudulent.

### False Positive

A **False Positive (FP)** occurs when the model predicts a positive result, but the actual result is negative.

Example:
A legitimate transaction is incorrectly classified as fraudulent.

### False Negative

A **False Negative (FN)** occurs when the model predicts a negative result, but the actual result is positive.

Example:
A fraudulent transaction is incorrectly classified as legitimate.

### True Negative

A **True Negative (TN)** occurs when the model correctly predicts a negative instance.

Example:
A legitimate transaction is correctly classified as legitimate.

---

## Precision

### Definition

Precision measures how many of the observations predicted as positive were actually positive.

It answers the question:

> When the model predicts a positive result, how often is it correct?

### Formula

\[
\text{Precision} = \frac{TP}{TP + FP}
\]

Where:

- `TP` = True Positives
- `FP` = False Positives

A high precision score means that the model produces relatively few false positive predictions.

### Example

If a model predicts that 60 transactions are fraudulent and 50 of those transactions are actually fraudulent:

\[
\text{Precision} = \frac{50}{50 + 10}
\]

\[
\text{Precision} = \frac{50}{60} = 0.83
\]

The model has a precision of approximately `0.83`, or `83%`.

---

## Recall

### Definition

Recall measures how many of the actual positive observations were correctly identified by the model.

It answers the question:

> Of all the actual positive cases, how many did the model find?

### Formula

\[
\text{Recall} = \frac{TP}{TP + FN}
\]

Where:

- `TP` = True Positives
- `FN` = False Negatives

A high recall score means that the model successfully identifies most of the actual positive cases.

### Example

If there are 55 actual fraudulent transactions and the model correctly identifies 50 of them:

\[
\text{Recall} = \frac{50}{50 + 5}
\]

\[
\text{Recall} = \frac{50}{55} = 0.91
\]

The model has a recall of approximately `0.91`, or `91%`.

---

## Precision vs. Recall

Precision and recall measure different aspects of model performance.

| Metric | Main Question | Penalizes |
|---|---|---|
| Precision | How many predicted positives were correct? | False positives |
| Recall | How many actual positives were detected? | False negatives |

### Precision is especially important when:
- False positives are expensive.
- Incorrect alerts waste time or resources.
- Legitimate transactions should not be blocked.
- Too many false alarms reduce trust in the system.

### Recall is especially important when:
- Missing a positive case is dangerous.
- Detecting as many positive cases as possible is important.
- Fraudulent transactions should not be missed.
- A medical condition must be detected early.

A model can have higher precision but lower recall, or higher recall but lower precision. The best metric depends on the problem being solved.

## Quiz: Comparing Precision and Recall
### Scenario:
You are evaluating two machine learning models based on their performance metrics. Model 1 has 50 true positives, 10 false positives, and 5 false negatives. Model 2 has 45 true positives, 5 false positives, and 10 false negatives.

### Tasks:

1. Calculate the precision and recall for both models.
2. Write code that gives a boolean value True if Model 1 has higher precision than Model 2, and False otherwise.
3. Write code that gives a boolean value True if Model 1 has higher recall than Model 2, and False otherwise.

````python
# Model 1 metrics
model1_tp = 50
model1_fp = 10
model1_fn = 5

# Model 2 metrics
model2_tp = 45
model2_fp = 5
model2_fn = 10

# Calculate precision and recall for Model 1
model1_precision = model1_tp / (model1_tp + model1_fp)
model1_recall = model1_tp / (model1_tp + model1_fn)

# Calculate precision and recall for Model 2
model2_precision = model2_tp / (model2_tp + model2_fp)
model2_recall = model2_tp / (model2_tp + model2_fn)

# Compare precision
precision_comparison_result = model1_precision > model2_precision

# Compare recall
recall_comparison_result = model1_recall > model2_recall

# Output the results
print(f"Model 1 Precision: {model1_precision:.2f}")
print(f"Model 1 Recall: {model1_recall:.2f}")
print(f"Model 2 Precision: {model2_precision:.2f}")
print(f"Model 2 Recall: {model2_recall:.2f}")
print(f"Is Model 1 precision higher than Model 2? {precision_comparison_result}")
print(f"Is Model 1 recall higher than Model 2? {recall_comparison_result}")
````

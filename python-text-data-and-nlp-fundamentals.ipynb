# 1. Working with Strings in Python

A **string** is a sequence of characters used to represent text.

In Python, strings can be created using single or double quotation marks.

```python
name = "William"
message = 'Hello, world!'
```

Quotation marks tell Python where the string begins and ends.

When a string itself contains quotation marks or apostrophes, the choice of quotation marks becomes important.

For example:

```python
quote = "You're doing great!"
```

The apostrophe in `You're` does not cause a problem because the string is enclosed in double quotation marks.

Similarly:

```python
quote = 'He said "Hello".'
```

uses single quotation marks around a string containing double quotation marks.

Correctly handling quotation marks is important when working with text because incorrectly formatted strings can result in Python syntax errors.

# 3. String Formatting and Logging

Machine learning and NLP systems often need to record information about their processing.

For example, an application may need to record:

* Which user made a request.

```python

```

The expressions inside `{}` are replaced with the values stored in the corresponding variables.

This is useful for creating readable and dynamic messages.

# 5. String Concatenation

**String concatenation** means joining two or more strings together.

Python uses the `+` operator to concatenate strings.

For example:

```python
first_name = "William"
last_name = "Pitt"

full_name = first_name + " " + last_name
```

The result is:

```text
William Pitt
```

The `" "` between the two variables represents a space.

Without the space:

```python
full_name = first_name + last_name
```

the result would be:

```text
WilliamPitt
```

When preparing text data, correctly handling spaces and other characters is important because formatting can affect how the text is interpreted.

# 6. Measuring Text Length

Python provides the `len()` function to determine the number of characters in a string.

For example:

```python
name = "William"

name_length = len(name)
```

The result is:

```text
7
```

Spaces are also characters and therefore count toward the length of a string.

For example:

```python
full_name = "William Pitt"

len(full_name)
```

returns:

```text
12
```

This includes the space between the first and last names.

Understanding string length is useful when validating text against limits imposed by applications, databases, APIs, or NLP processing pipelines.

# 7. Named Entity Recognition (NER)

**Named Entity Recognition (NER)** is an NLP task that identifies and classifies important entities within text.

Depending on the NLP system, entities may include:

* People
* Organizations
* Locations
* Dates
* Times
* Monetary values
* Percentages
* Quantities

For example, given:

```text
William Bradley Pitt was born in Oklahoma.
```

an NER system could identify:

```text
William Bradley Pitt → PERSON
Oklahoma → LOCATION
```

NER is commonly used in information extraction and document-processing systems.

For example, an organization could use NER to extract:

* Customer names
* Company names
* Locations
* Dates
* Invoice numbers
* Financial amounts

from large collections of documents.

# 8. Preparing Text for NER

Before text is processed by an NLP model, it may need to satisfy certain requirements.

For example, an application may impose a maximum character limit on a particular field.

Python can be used to validate the text before sending it to an NLP pipeline.

```python
given_name = "William"
middle_names = "Bradley"
family_name = "Pitt"

full_name = given_name + " " + middle_names + " " + family_name

name_length = len(full_name)
```

The resulting length can then be compared against a limit:

```python
named_entity_character_limit = 28

print(name_length <= named_entity_character_limit)
```

The comparison produces a Boolean value:

```text
True
```

or:

```text
False
```

This demonstrates how simple Python operations can be used as part of data validation before NLP processing.

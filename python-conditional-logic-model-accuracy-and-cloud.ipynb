# Exercise 1: Model Accuracy Guess

## 1. Exercise Overview

The first exercise compares a predicted model accuracy with the actual model accuracy.

The program receives two values:

```python
actual_accuracy = 0.85
predicted_accuracy = 0.80
```

The goal is to determine whether the prediction was:

* Too low.
* Too high.
* Exactly correct.

This exercise introduces a simple comparison between two numerical values.

---

## 2. Understanding the Inputs

The actual accuracy represents the model's measured performance.

```python
actual_accuracy = 0.85
```

The predicted accuracy represents the value provided by another user.

```python
predicted_accuracy = 0.80
```

In this example:

```text
Actual accuracy:    0.85
Predicted accuracy: 0.80
```

Since:

```text
0.80 < 0.85
```

the prediction is too low.

The expected result is:

```text
Oops! Your prediction was too low.
```

---

## 3. Required Decision Logic

The program needs to compare:

```python
predicted_accuracy
```

with:

```python
actual_accuracy
```

The correct decision structure is:

```python
if predicted_accuracy < actual_accuracy:
    # prediction is too low
elif predicted_accuracy > actual_accuracy:
    # prediction is too high
else:
    # prediction matches actual accuracy
```

This structure handles all three possible outcomes.

### Case 1: Prediction is too low

```python
predicted_accuracy < actual_accuracy
```

Example:

```python
0.80 < 0.85
```

Result:

```text
Oops! Your prediction was too low.
```

### Case 2: Prediction is too high

```python
predicted_accuracy > actual_accuracy
```

Example:

```python
0.90 > 0.85
```

Result:

```text
Oops! Your prediction was too high.
```

### Case 3: Prediction matches

If neither condition is true, the values must be equal.

Example:

```python
0.85 == 0.85
```

Result:

```text
Nice! Your prediction matched the actual accuracy!
```

---

## 4. Important Correction in the Starter Code

The starter code contains this condition:

```python
if predicted_accuracy < predicted_accuracy:
```

This is incorrect because it compares the predicted accuracy with itself.

A value cannot normally be less than itself.

For example:

```python
0.80 < 0.80
```

evaluates to:

```python
False
```

The condition should compare the prediction with the actual accuracy:

```python
if predicted_accuracy < actual_accuracy:
```

This is an important debugging lesson:

> When comparing two pieces of information, make sure each side of the comparison refers to the correct variable.

---

## 5. Complete Solution

```python
# Actual and predicted accuracies

actual_accuracy = 0.85
predicted_accuracy = 0.80

# Compare predicted accuracy to actual accuracy

if predicted_accuracy < actual_accuracy:
    result = "Oops! Your prediction was too low."
elif predicted_accuracy > actual_accuracy:
    result = "Oops! Your prediction was too high."
else:
    result = "Nice! Your prediction matched the actual accuracy!"

# Check the result

print(result)
```

### Expected Output

```text
Oops! Your prediction was too low.
```

---

## 6. How the Code Works

Python evaluates the first condition:

```python
predicted_accuracy < actual_accuracy
```

Using the example values, this becomes:

```python
0.80 < 0.85
```

The result is:

```python
True
```

Therefore, Python executes:

```python
result = "Oops! Your prediction was too low."
```

Because the first condition is true, Python skips the remaining `elif` and `else` branches.

The result is then displayed using:

```python
print(result)
```

# Exercise 2: Compute Resource Cost Calculation

## 1. Exercise Overview

The second exercise applies conditional logic to cloud computing costs.

Different cloud providers use different cost rates:

| Provider | Cost Rate |
| -------- | --------: |
| AWS      |      7.5% |
| Azure    |      9.5% |
| GCP      |      8.9% |

The program receives:

```python
provider = "AWS"
computation_cost = 1000
```

It must:

1. Identify the selected provider.
2. Apply the correct cost rate.
3. Calculate the total cost.
4. Display a formatted message.

---

## 2. Understanding the Inputs

The provider is stored as a string:

```python
provider = "AWS"
```

The initial computation cost is stored as a number:

```python
computation_cost = 1000
```

The computation cost represents the original cost before applying the provider-specific rate.

For AWS:

```text
Initial computation cost = $1,000
AWS rate = 7.5%
```

The rate as a decimal is:

```python
0.075
```

---

## 3. Converting Percentages to Decimals

Percentages must be converted into decimal values before being used in Python calculations.

The conversion is:

```text
percentage ÷ 100
```

Therefore:

```text
7.5%  = 0.075
9.5%  = 0.095
8.9%  = 0.089
```

The rates can be represented as:

```python
aws_rate = 0.075
azure_rate = 0.095
gcp_rate = 0.089
```

---

## 4. Calculating the Total Cost

The total cost includes the original computation cost plus the additional provider rate.

The formula is:

```python
total_cost = computation_cost * (1 + cost_rate)
```

For AWS:

```python
total_cost = 1000 * (1 + 0.075)
```

This becomes:

```python
total_cost = 1000 * 1.075
```

Therefore:

```text
Total cost = $1,075.00
```

The `1` represents the original 100% of the cost.

The additional rate is added to that original amount.

---

## 5. Required Decision Logic

The provider determines which rate should be used.

The structure is:

```python
if provider == "AWS":
    cost_rate = 0.075
elif provider == "Azure":
    cost_rate = 0.095
elif provider == "GCP":
    cost_rate = 0.089
else:
    # provider is not recognized
```

This introduces an important concept:

> The program selects a calculation rule based on the value of a variable.

---

## 6. Complete Solution

```python
# Provider and computation cost

provider = "AWS"
computation_cost = 1000

# Determine the cost rate based on the provider
# and calculate the total cost

if provider == "AWS":
    cost_rate = 0.075
    total_cost = computation_cost * (1 + cost_rate)
    result = "Since you are using {}, your total cost is ${:.2f}.".format(
        provider,
        total_cost
    )

elif provider == "Azure":
    cost_rate = 0.095
    total_cost = computation_cost * (1 + cost_rate)
    result = "Since you are using {}, your total cost is ${:.2f}.".format(
        provider,
        total_cost
    )

elif provider == "GCP":
    cost_rate = 0.089
    total_cost = computation_cost * (1 + cost_rate)
    result = "Since you are using {}, your total cost is ${:.2f}.".format(
        provider,
        total_cost
    )

else:
    result = "Provider not recognized."

print(result)
```

### Expected Output

```text
Since you are using AWS, your total cost is $1075.00.
```

---

## 7. Understanding the Provider Conditions

The code uses equality comparison:

```python
provider == "AWS"
```

This checks whether the value stored in `provider` is equal to the string `"AWS"`.

It is important to use:

```python
==
```

rather than:

```python
=
```

Correct:

```python
if provider == "AWS":
```

Incorrect:

```python
if provider = "AWS":
```

The incorrect version attempts assignment inside the condition and produces a syntax error.

---

## 8. Understanding the Cost Calculation

For AWS:

```python
cost_rate = 0.075
```

The total cost is calculated using:

```python
total_cost = computation_cost * (1 + cost_rate)
```

With the example input:

```python
computation_cost = 1000
```

the calculation is:

```python
1000 * (1 + 0.075)
```

which produces:

```python
1075.0
```

The result is formatted to two decimal places:

```python
"{:.2f}"
```

This means:

* `f` → format the value.
* `.2` → show two digits after the decimal point.

Therefore:

```python
1075.0
```

is displayed as:

```text
1075.00
```

This is useful when displaying currency values.

---

## 9. Understanding `.format()`

The message uses:

```python
"Since you are using {}, your total cost is ${:.2f}.".format(
    provider,
    total_cost
)
```

The first `{}` is replaced with:

```python
provider
```

The second placeholder:

```python
{:.2f}
```

is replaced with the formatted total cost.

For example:

```python
provider = "AWS"
total_cost = 1075.0
```

produces:

```text
Since you are using AWS, your total cost is $1075.00.
```

This is an example of string formatting.

## 11. A More Efficient Version

The repeated calculation and message formatting can be simplified.

Instead of calculating the total cost separately inside every provider branch, I can first select the rate and then perform the calculation once.

```python
provider = "AWS"
computation_cost = 1000

if provider == "AWS":
    cost_rate = 0.075
elif provider == "Azure":
    cost_rate = 0.095
elif provider == "GCP":
    cost_rate = 0.089
else:
    cost_rate = None

if cost_rate is None:
    result = "Provider not recognized."
else:
    total_cost = computation_cost * (1 + cost_rate)
    result = "Since you are using {}, your total cost is ${:.2f}.".format(
        provider,
        total_cost
    )

print(result)
```

This version demonstrates better code organization because:

* Provider selection happens in one place.
* The calculation is written only once.
* The output formatting is written only once.
* The code is easier to maintain.

The original exercise may expect the longer structure, but this alternative is useful for understanding how code can be improved after the basic solution works.

## Exercise 1: Model Accuracy Guess

In this exercise, you'll compare a predicted accuracy to an actual accuracy and provide feedback on whether the predicted accuracy is too low, too high, or matches the actual accuracy.

### Problem Statement

You have an actual model accuracy stored in a variable called `actual_accuracy`. Another user provides a predicted accuracy called `predicted_accuracy`. By comparing `predicted_accuracy` to `actual_accuracy`, inform the user if their prediction is too high, too low, or correct.

**Example Input**:
```python
actual_accuracy = 0.85  # replace with actual accuracy
predicted_accuracy = 0.80  # replace with predicted accuracy
```

**Instructions**:
1. Compare `predicted_accuracy` to `actual_accuracy`.
2. Inform the user if their prediction is too high, too low, or correct.

````python
# Actual and predicted accuracies
actual_accuracy = 0.85  # replace with actual accuracy
predicted_accuracy = 0.80  # replace with predicted accuracy

# Compare predicted accuracy to actual accuracy
# TODO
if predicted_accuracy < actual_accuracy:
    result = "Oops! Your prediction was too low."
elif predicted_accuracy > actual_accuracy:
    result = "Oops! Your prediction was too high."
else:
    result = "Nice! Your prediction matched the actual accuracy!"

# Check the result
print(result)

### Notebook grading
def get_solution(actual_accuracy, predicted_accuracy):
    if predicted_accuracy < actual_accuracy:
        return "Oops! Your prediction was too low."
    elif predicted_accuracy > actual_accuracy:
        return "Oops! Your prediction was too high."
    else:
        return "Nice! Your prediction matched the actual accuracy!"

if result == get_solution(actual_accuracy, predicted_accuracy):
    print("Good job!")
else:
    print("Try again. That doesn't look like the expected answer.")
````

## Exercise: Compute Resource Cost Calculation

In this exercise, you'll calculate the total cost of running a machine learning model training session based on the cloud provider used. Different providers have different cost rates.

### Problem Statement

Depending on the cloud provider, you need to apply the appropriate cost rate to the computation cost. The providers and their rates are as follows: AWS (7.5%), Azure (9.5%), and GCP (8.9%). Use this information to calculate the total cost of running the training session based on the provider and the initial computation cost.

**Example Input**:
```python
provider = "AWS"  # Either "AWS", "Azure", or "GCP"
computation_cost = 1000  # amount of computation cost
```

**Instructions**:
1. Use the provider to determine the cost rate.
2. Calculate the total cost based on the computation cost and the cost rate.
3. Inform the user of the total cost based on their provider.

````python
# Provider and computation cost
provider = "GCP"  # Either "AWS", "Azure", or "GCP"
computation_cost = 1000  # amount of computation cost

# Determine the cost rate based on the provider and calculate the total cost
if provider == "AWS":
    cost_rate = 0.075
    total_cost = computation_cost * (1 + cost_rate)
    result = "Since you are using {}, your total cost is ${:.2f}.".format(
        provider, total_cost
    )
elif provider == "Azure":
    cost_rate = 0.095
    total_cost = computation_cost * (1 + cost_rate)
    result = "Since you are using {}, your total cost is ${:.2f}.". format(
        provider, total_cost
    )
elif provider == "GCP":
    cost_rate = 0.089
    total_cost = computation_cost * (1 + cost_rate)
    result = "Since you are using {}, your total cost is ${:.2f}.". format(
        provider, total_cost
    )
else:
    result = "Provider not recognized."
    
# Check the result
print(result)
### Notebook grading
def get_solution(provider, computation_cost):
    if provider == 'AWS':
        cost_rate = .075
        total_cost = computation_cost * (1 + cost_rate)
        result = "Since you are using {}, your total cost is ${:.2f}.".format(provider, total_cost)
    elif provider == 'Azure':
        cost_rate = .095
        total_cost = computation_cost * (1 + cost_rate)
        result = "Since you are using {}, your total cost is ${:.2f}.".format(provider, total_cost)
    elif provider == 'GCP':
        cost_rate = .089
        total_cost = computation_cost * (1 + cost_rate)
        result = "Since you are using {}, your total cost is ${:.2f}.".format(provider, total_cost)
    else:
        result = "Provider not recognized."
    return result

if result == get_solution(provider, computation_cost):
    print("Good job!")
else:
    print("Oops! That doesn't look like the expected answer.")
````
